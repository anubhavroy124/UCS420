```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.plot(x, y1, color='red', linestyle='-', label='y = M*x^2')
plt.plot(x, y2, color='blue', linestyle='--', label='y = M*sin(x)')

plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
data = {
    'Subject': ['Maths', 'Physics', 'Chemistry', 'Computer Science', 'English'],
    'Score': [85, 78, 92, 95, 80]
}

df = pd.DataFrame(data)

print(df)

sns.barplot(data=df, x='Subject', y='Score', hue='Subject', palette='Set2', legend=False)

for i, score in enumerate(df['Score']):
    plt.text(i, score + 1, str(score), ha='center')

plt.title("Scores in Five Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
np.random.seed(1024170150)

data = np.random.randn(50)
x = np.arange(1, 51)

cumulative_sum = np.cumsum(data)
noise = data + np.random.randn(50) * 0.5

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

ax[0, 0].plot(x, cumulative_sum, color='blue')
ax[0, 0].set_title("Cumulative Sum")
ax[0, 0].set_xlabel("Index")
ax[0, 0].set_ylabel("Cumulative Sum")
ax[0, 0].grid(True)

ax[0, 1].scatter(x, noise, color='red')
ax[0, 1].set_title("Scatter Plot with Random Noise")
ax[0, 1].set_xlabel("Index")
ax[0, 1].set_ylabel("Value")
ax[0, 1].grid(True)

ax[1, 0].plot(x, data, color='green')
ax[1, 0].set_title("Random Data Line Plot")
ax[1, 0].set_xlabel("Index")
ax[1, 0].set_ylabel("Random Value")
ax[1, 0].grid(True)

ax[1, 1].hist(data, bins=10, color='orange', edgecolor='black')
ax[1, 1].set_title("Distribution of Random Data")
ax[1, 1].set_xlabel("Value")
ax[1, 1].set_ylabel("Frequency")
ax[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

df = pd.read_csv(url)

print(df.head())

sns.lineplot(data=df, x='month_number', y='total_profit', marker='o')

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid(True)
plt.show()

product_columns = [
    'facecream', 'facewash', 'toothpaste',
    'bathingsoap', 'shampoo', 'moisturizer'
]

for col in product_columns:
    sns.lineplot(data=df, x='month_number', y=col, marker='o', label=col)

plt.title("Monthly Sales of All Products")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True)
plt.show()

for col in df.columns:
    plt.figure(figsize=(8, 4))
    sns.barplot(data=df, x='month_number', y=col, color='skyblue')
    plt.title("Bar Chart of " + col)
    plt.xlabel("Month Number")
    plt.ylabel(col)
    plt.grid(axis='y')
    plt.show()
```
