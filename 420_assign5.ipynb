Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic
operations like:
a) Addition of 2 in all the element
b) Multiply 3 with all the elements
c) Divide every element by 2

```python
arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)
add_result = arr + 2
print("After adding 2:", add_result)
mul_result = arr * 3
print("After multiplying by 3:", mul_result)
divide_by_2 = arr / 2
print("After division: ", divide_by_2)
```

Q.2 Questions on Basic NumPy Array:
a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Reversed array:", reversed_arr)
def most_frequent(arr):
    values, counts = np.unique(arr, return_counts=True)
    max_count = counts.max()
    most_frequent_values = values[counts == max_count]
    indices = np.where(np.isin(arr, most_frequent_values))[0]
    return most_frequent_values, indices

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

x_val, x_idx = most_frequent(x)
print("x -> Most frequent value(s):", x_val, "at indices:", x_idx)

y_val, y_idx = most_frequent(y)
print("y -> Most frequent value(s):", y_val, "at indices:", y_idx)
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
Anubhav = np.linspace(10, 100, 25)

print("Array:", Anubhav)
print("Dimensions:", Anubhav.ndim)
print("Shape:", Anubhav.shape)
print("Total elements:", Anubhav.size)
print("Data type:", Anubhav.dtype)
print("Total bytes:", Anubhav.nbytes)

reshaped_Anubhav = Anubhav.reshape(1, 25)

print("Transposed array using reshape():")
print(reshaped_Anubhav)

print("Transpose using T attribute:")
print(Anubhav.T)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_Anubhav = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])

print("Original array:")
print(ucs420_Anubhav)

print("Mean:", np.mean(ucs420_Anubhav))
print("Median:", np.median(ucs420_Anubhav))
print("Maximum:", np.max(ucs420_Anubhav))
print("Minimum:", np.min(ucs420_Anubhav))
print("Unique elements:", np.unique(ucs420_Anubhav))

reshaped_ucs420_Anubhav = ucs420_Anubhav.reshape(4, 3)

print("Reshaped array (4 rows, 3 columns):")
print(reshaped_ucs420_Anubhav)

resized_ucs420_Anubhav = np.resize(ucs420_Anubhav, (2, 3))

print("Resized array (2 rows, 3 columns):")
print(resized_ucs420_Anubhav)
```
