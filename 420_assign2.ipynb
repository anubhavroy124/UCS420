```python
roll_no = "1024170150"

L = [int(digit) * 10 for digit in roll_no]

print("L =", L)

L.append(500)
print("After append:", L)

# append() adds 500 at the end of the list

L.insert(2, 700)
print("After insert:", L)

# insert() adds 700 at index 2 and shifts the other elements

L.remove(500)
print("After remove:", L)

L.pop(2)
print("After pop:", L)

L.sort()
print("Ascending order:", L)

L.sort(reverse=True)
print("Descending order:", L)

print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

average = sum(L) / len(L)
new_list = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average:", new_list)
```

```python
scores = tuple(L[:8])

print("Scores tuple:", scores)

highest = max(scores)
highest_index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Lowest score:", lowest)
print("Number of times lowest appears:", lowest_count)

reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)

x = int(input("Enter a score to search: "))

if x in scores:
    print("First occurrence index:", scores.index(x))
else:
    print("Score not present in tuple")



first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
scores[0] = 100
```

```python
import random

random.seed(1024170150)

numbers = [random.randint(100, 900) for i in range(100)]

print("Random numbers:")
print(numbers)

odd_numbers = [x for x in numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Number of odd numbers:", len(odd_numbers))

even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
print("Number of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

most_frequent = max(numbers, key=numbers.count)
frequency = numbers.count(most_frequent)

print("Most frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
roll_no = "1024170150"

digits = [int(digit) for digit in roll_no[:8]]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

union = A.union(B)
print("Union:", union)

intersection = A.intersection(B)
print("Intersection:", intersection)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

symmetric = A.symmetric_difference(B)
print("Symmetric difference:", symmetric)

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

X = int(input("Enter value X: "))

A.discard(X)

print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Anubhav",
    "roll_no": "1024170150",
    "branch": "CSE",
    "age": 19,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")

print("After renaming city:")
print(my_dict)

my_dict["cgpa"] = 8.5

print("After adding CGPA:")
print(my_dict)

my_dict["age"] = my_dict["age"] + 1

print("After increasing age:")
print(my_dict)

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

print("Using pop():")
print(dict_pop)
print("Removed value:", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]

print("Using del:")
print(dict_del)

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email not found in dictionary")

friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170151",
    "branch": "CSE",
    "age": 19,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:")
print(merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("String values dictionary:")
print(string_values)
```
