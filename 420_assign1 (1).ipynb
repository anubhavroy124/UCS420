```python
n = int(input("Enter number: "))
s = 0

for i in range(2, n + 1):
    flag = 0

    for j in range(2, i // 2 + 1):
        if i % j == 0:
            flag = 1
            break

    if flag == 0:
        s = s + i

print("Sum is", s)
```

```python
def oddsum(n):
    s = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i
    return s

n = int(input("Enter number: "))
print("Sum is", oddsum(n))
```

```python
def primesum(n):
    s = 0

    for i in range(2, n + 1):
        flag = 0

        for j in range(2, i // 2 + 1):
            if i % j == 0:
                flag = 1
                break

        if flag == 0:
            s = s + i

    return s


n = int(input("Enter number: "))
print("Sum is", primesum(n))
```
